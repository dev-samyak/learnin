## 1. Arrays and Strings

An array is a collection of elements stored in contiguous memory locations. Arrays allow fast indexing, but their size is often fixed in lower-level languages.

A string is a sequence of characters, which can be treated as an array-like structure in many languages.

### Key ideas
- Static arrays have fixed size.
- Dynamic arrays can grow as needed.
- Multidimensional arrays store data in rows and columns.
- Strings support indexing, slicing, concatenation, and searching.

```python
# Arrays and strings in Python
arr = [10, 20, 30, 40]
print(arr[0])
print(arr[-1])
print(arr[1:3])

text = "data structures"
print(text.upper())
print(text.split())
```

## 2. Linked Lists

A linked list stores data in nodes. Each node contains data and a pointer to the next node.

### Types
- Singly linked list: each node points to the next node
- Doubly linked list: each node points to both previous and next nodes
- Circular linked list: the last node points back to the first

### Common operations
- Insert at the beginning, middle, or end
- Delete a node
- Traverse the list

```python
class Node:
    def __init__(self, value):
        self.value = value
        self.next = None

head = Node(10)
head.next = Node(20)
head.next.next = Node(30)

current = head
while current:
    print(current.value, end=" -> ")
    current = current.next
print("None")
```

## 3. Stacks

A stack follows the Last In, First Out (LIFO) rule. The most recently added element is removed first.

### Common operations
- Push: add an element
- Pop: remove the top element
- Peek: view the top element

### Use cases
- Expression evaluation
- Undo/redo operations
- Recursion simulation

```python
stack = []
stack.append(10)
stack.append(20)
stack.append(30)
print(stack)
print(stack.pop())
print(stack)
```

## 4. Queues

A queue follows the First In, First Out (FIFO) rule. The first added element is removed first.

### Variants
- Simple queue
- Circular queue
- Deque (double-ended queue)
- Priority queue

### Use cases
- Task scheduling
- Buffering data
- Printer queues

```python
from collections import deque

q = deque([10, 20, 30])
q.append(40)
print(q.popleft())
print(q)
```

### Important terms

### Traversal methods

## 6. Binary Search Trees

A binary search tree (BST) is a binary tree where the left subtree contains smaller values and the right subtree contains larger values.

### Common operations
- Search
- Insert
- Delete
- Traversal

### Notes
- BSTs are efficient for lookup when balanced.
- Unbalanced BSTs can degrade to linked-list-like performance.

```python
class BSTNode:
    def __init__(self, value):
        self.value = value
        self.left = None
        self.right = None


def insert(root, value):
    if root is None:
        return BSTNode(value)
    if value < root.value:
        root.left = insert(root.left, value)
    else:
        root.right = insert(root.right, value)
    return root

root = None
for x in [50, 20, 70, 10, 30]:
    root = insert(root, x)

print(root.value)
print(root.left.value)
```

## 7. Heaps and Priority Queues

A heap is a special tree-based structure that satisfies the heap property. In a min-heap, the smallest element is at the root. In a max-heap, the largest element is at the root.

### Applications
- Priority queues
- Scheduling tasks
- Heap sort

```python
import heapq

heap = [4, 10, 7, 1, 3]
heapq.heapify(heap)
print(heap)
heapq.heappush(heap, 2)
print(heap)
print(heapq.heappop(heap))
```

## 8. Hashing and Hash Tables

A hash table stores data using a hash function that maps keys to indexes. It provides average $O(1)$ time for insertion, deletion, and lookup.

### Important concepts
- Hash function
- Collision handling
- Chaining and open addressing

### Trade-offs
- Fast average case
- Worst-case performance can be poor if many collisions occur

```python
student_marks = {"Alice": 90, "Bob": 85, "Charlie": 88}
print(student_marks["Bob"])
student_marks["Diana"] = 92
print(student_marks)
```

## 9. Graphs

A graph is a collection of vertices connected by edges. Graphs can be directed or undirected.

### Representations
- Adjacency matrix
- Adjacency list

### Common algorithms
- Breadth-first search (BFS)
- Depth-first search (DFS)
- Shortest path algorithms

```python
graph = {
    'A': ['B', 'C'],
    'B': ['D'],
    'C': ['E'],
    'D': [],
    'E': []
}

print(graph['A'])
```

## 10. Sorting Algorithms

Sorting arranges elements in a particular order, usually ascending or descending.

### Common methods
- Bubble sort
- Selection sort
- Insertion sort
- Merge sort
- Quick sort

### Key idea
Different algorithms have different time complexities and are suitable for different data sizes.

```python
arr = [5, 2, 9, 1, 7]
print(sorted(arr))
print(arr)
```

## 11. Searching Algorithms

Searching finds whether a value exists in a collection and often returns its position.

### Methods
- Linear search: check each element one by one
- Binary search: repeatedly divide the search space in half

### Note
Binary search works only on sorted data.

```python
def linear_search(values, target):
    for i, value in enumerate(values):
        if value == target:
            return i
    return -1

values = [10, 20, 30, 40, 50]
print(linear_search(values, 30))
```

## 12. Complexity and Performance

Time complexity describes how the running time grows as input size increases. Space complexity describes how much extra memory is needed.

### Common Big O notations
- $O(1)$: constant time
- $O(log n)$: logarithmic time
- $O(n)$: linear time
- $O(n log n)$: linearithmic time
- $O(n^2)$: quadratic time
