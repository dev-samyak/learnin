## 7. Advanced Functions

Python supports higher-order functions, lambda expressions, and flexible argument handling.

````python
def apply_operation(values, func):
    return [func(v) for v in values]

numbers = [1, 2, 3, 4]
print(apply_operation(numbers, lambda x: x * x))

def show_details(*args, **kwargs):
    print(args)
    print(kwargs)

show_details(1, 2, name='Alice', active=True)
````

## 8. Comprehensions and Generators

Comprehensions provide concise ways to create collections, while generators are memory-efficient for large data.

````python
squares = [x ** 2 for x in range(10)]
print(squares)

even_squares = {x: x ** 2 for x in range(6) if x % 2 == 0}
print(even_squares)

def number_generator(limit):
    for i in range(limit):
        yield i

for value in number_generator(5):
    print(value)
````

## 9. Object-Oriented Programming

Classes and inheritance allow you to model real-world concepts in a structured way.

````python
class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        return f'{self.name} makes a sound'

class Dog(Animal):
    def speak(self):
        return f'{self.name} barks'

print(Dog('Buddy').speak())
````

## 10. Decorators and Context Managers

Decorators wrap functions, and context managers help manage setup and cleanup logic.

````python
def log_calls(func):
    def wrapper(*args, **kwargs):
        print(f'Calling {func.__name__}')
        return func(*args, **kwargs)
    return wrapper

@log_calls
def greet_user(name):
    return f'Hello, {name}'

print(greet_user('Emma'))

from contextlib import contextmanager

@contextmanager
def managed_resource():
    print('Acquiring resource')
    try:
        yield
    finally:
        print('Releasing resource')

with managed_resource():
    print('Using resource')
````

## 11. Type Hints, Dataclasses, and Modules

Type hints improve readability, while dataclasses make structured data cleaner.

````python
from dataclasses import dataclass
from typing import List

@dataclass
class Point:
    x: int
    y: int

def distance(points: List[Point]) -> float:
    return sum((p.x + p.y) for p in points)

print(distance([Point(1, 2), Point(3, 4)]))
````

## 12. Asynchronous Programming

Async and await are used for I/O-bound tasks such as API calls or file operations.

````python
import asyncio

async def worker(name):
    await asyncio.sleep(0.1)
    print(f'Done {name}')

async def main():
    await asyncio.gather(worker('A'), worker('B'))

asyncio.run(main())
````

## 14. Concurrency and Parallelism

Thread pools are useful for I/O-bound tasks, while multiprocessing helps with CPU-bound work.

````python
from concurrent.futures import ThreadPoolExecutor
import time

def task(name, delay):
    time.sleep(delay)
    return f'{name} completed'

with ThreadPoolExecutor(max_workers=3) as executor:
    futures = [executor.submit(task, f'task-{i}', i * 0.1) for i in range(3)]
    for future in futures:
        print(future.result())
````

## 15. Descriptors and Advanced OOP

Descriptors let objects customize attribute access and enforce behavior.

````python
class Descriptor:
    def __get__(self, instance, owner):
        return 42

    def __set__(self, instance, value):
        instance._value = value

class Example:
    attr = Descriptor()

obj = Example()
print(obj.attr)
obj.attr = 10
print(obj._value)
````

## 16. Profiling and Performance

Profiling tools help identify bottlenecks and improve runtime efficiency.

````python
import cProfile
import pstats
import io

def expensive_work():
    total = 0
    for i in range(10000):
        total += i * i
    return total

profiler = cProfile.Profile()
profiler.enable()
expensive_work()
profiler.disable()

stream = io.StringIO()
stats = pstats.Stats(profiler, stream=stream).sort_stats('cumulative')
stats.print_stats(5)
print(stream.getvalue())
````

## 28. Observability and Structured Logging

Production systems need correlated logs, clear error handling, and useful context.

````python
import logging
import contextvars

request_id = contextvars.ContextVar("request_id", default="n/a")

def log_event(message: str):
    logger = logging.getLogger("app")
    logger.setLevel(logging.INFO)
    if not logger.handlers:
        logging.basicConfig(level=logging.INFO, format="%(asctime)s | %(levelname)s | %(message)s")
    logger.info("event=%s request_id=%s", message, request_id.get())

request_id.set("req-42")
log_event("user_login")
````

## 30. Message Queues and Background Work

Queues help decouple services and make workflows more resilient and scalable.

````python
from queue import Queue
import threading

q = Queue()

def worker():
    while not q.empty():
        item = q.get()
        print(f"processed: {item}")
        q.task_done()

for item in ["email", "report", "sync"]:
    q.put(item)

threads = [threading.Thread(target=worker) for _ in range(2)]
for t in threads:
    t.start()
for t in threads:
    t.join()
q.join()
````

## 33. Architecture and Design Patterns

Design patterns are reusable solutions to common software design problems. They help improve maintainability, flexibility, readability, and scalability in real-world systems.

There are three major families:
- Creational patterns: used when creating objects in a flexible and controlled way
- Structural patterns: used when organizing classes and objects into larger, useful structures
- Behavioral patterns: used when defining communication and responsibility between objects

Common patterns and where they are used:

1. Factory Pattern
   - Used when you want to create objects without exposing the exact class being instantiated
   - Common in plugin systems, object creation based on input, and service factories

2. Builder Pattern
   - Used when an object has many optional parameters or complex construction steps
   - Common in API clients, configuration objects, and report builders

3. Singleton Pattern
   - Used when one shared instance of a class should exist across the app
   - Common in configuration managers, logging services, and connection pools

4. Prototype Pattern
   - Used when creating a copy of an existing object is more efficient than rebuilding it
   - Common in cloning objects, caching, and document-like structures

5. Adapter Pattern
   - Used when two incompatible interfaces need to work together
   - Common in third-party integrations, legacy system bridges, and SDK wrappers

6. Facade Pattern
   - Used to simplify a complex subsystem behind a clean interface
   - Common in libraries, service layers, and API gateways

7. Decorator Pattern
   - Used to add behavior dynamically without changing the original class
   - Common in logging, authentication, validation, and caching wrappers

8. Proxy Pattern
   - Used to control access to another object, often for lazy loading or security
   - Common in remote services, access control, and lazy-loaded resources

9. Strategy Pattern
   - Used when multiple algorithms or behaviors can be selected at runtime
   - Common in payment systems, sorting logic, and routing rules

10. Observer Pattern
   - Used when one object needs to notify many dependents about changes
   - Common in event systems, UI updates, and pub/sub architectures

11. Command Pattern
   - Used to encapsulate a request as an object so it can be queued, logged, or executed later
   - Common in task queues, undo/redo systems, and workflow engines

12. State Pattern
   - Used when an object behaves differently depending on its current state
   - Common in order processing, workflow engines, and UI state machines

13. Template Method Pattern
   - Used when multiple classes share a common algorithm with small variations
   - Common in data processing pipelines, framework hooks, and test suites

14. Iterator Pattern
   - Used to traverse a collection without exposing its internal structure
   - Common in lists, trees, custom containers, and streaming data

15. Chain of Responsibility Pattern
   - Used when multiple handlers may process a request in sequence
   - Common in middleware, validation chains, and logging pipelines

A good senior developer uses patterns when they solve a recurring problem clearly, but avoids over-engineering simple code.

````python
from abc import ABC, abstractmethod

# Factory Pattern
class Dog:
    def speak(self):
        return "woof"

class Cat:
    def speak(self):
        return "meow"

class AnimalFactory:
    def create(self, animal_type: str):
        if animal_type == "dog":
            return Dog()
        if animal_type == "cat":
            return Cat()
        raise ValueError("Unknown animal")

factory = AnimalFactory()
print(factory.create("dog").speak())

# Strategy Pattern
class PaymentStrategy(ABC):
    @abstractmethod
    def pay(self, amount: float) -> str:
        pass

class CardPayment(PaymentStrategy):
    def pay(self, amount: float) -> str:
        return f"Paid {amount} with card"

class CashPayment(PaymentStrategy):
    def pay(self, amount: float) -> str:
        return f"Paid {amount} with cash"

def checkout(strategy: PaymentStrategy, amount: float):
    return strategy.pay(amount)

print(checkout(CardPayment(), 100))

# Decorator Pattern
def log_calls(func):
    def wrapper(*args, **kwargs):
        print(f"Calling {func.__name__}")
        return func(*args, **kwargs)
    return wrapper

@log_calls
def greet(name: str) -> str:
    return f"Hello {name}"

print(greet("Alice"))

# Singleton-style example
class Config:
    _instance = None

    def __new__(cls):
        if cls._instance is None:
            cls._instance = super().__new__(cls)
        return cls._instance

cfg1 = Config()
cfg2 = Config()
print(cfg1 is cfg2)
````
