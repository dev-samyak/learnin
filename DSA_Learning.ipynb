## 1. Big-O Notation

Big-O describes how the runtime or memory usage of an algorithm grows as the input size increases.

- O(1): Constant time
- O(log n): Logarithmic time
- O(n): Linear time
- O(n log n): Efficient sorting
- O(n^2): Quadratic time

### Python Example
```python
def find_first(arr, target):
    for i, value in enumerate(arr):
        if value == target:
            return i
    return -1

print(find_first([3, 5, 7, 9], 7))
```

### Java Example
```java
public class Main {
    public static int findFirst(int[] arr, int target) {
        for (int i = 0; i < arr.length; i++) {
            if (arr[i] == target) return i;
        }
        return -1;
    }

    public static void main(String[] args) {
        int[] arr = {3, 5, 7, 9};
        System.out.println(findFirst(arr, 7));
    }
}
```

## 7. Recursion

Recursion solves a problem by breaking it into smaller versions of itself.

### Python Example
```python
def factorial(n):
    if n <= 1:
        return 1
    return n * factorial(n - 1)

print(factorial(5))
```

### Java Example
```java
public class Main {
    public static int factorial(int n) {
        if (n <= 1) return 1;
        return n * factorial(n - 1);
    }

    public static void main(String[] args) {
        System.out.println(factorial(5));
    }
}
```

## 8. Sorting and Searching

Searching and sorting are foundational algorithms. Binary search is fast for sorted data, while merge sort and quick sort are common sorting methods.

### Python Example
```python
def binary_search(arr, target):
    left, right = 0, len(arr) - 1
    while left <= right:
        mid = (left + right) // 2
        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            left = mid + 1
        else:
            right = mid - 1
    return -1

print(binary_search([1, 3, 5, 7, 9], 7))
```

### Java Example
```java
public class Main {
    public static int binarySearch(int[] arr, int target) {
        int left = 0, right = arr.length - 1;
        while (left <= right) {
            int mid = left + (right - left) / 2;
            if (arr[mid] == target) return mid;
            if (arr[mid] < target) left = mid + 1;
            else right = mid - 1;
        }
        return -1;
    }

    public static void main(String[] args) {
        int[] arr = {1, 3, 5, 7, 9};
        System.out.println(binarySearch(arr, 7));
    }
}
```

## 9. Graphs and Traversal

Graphs model relationships between entities. Common traversal techniques include Breadth-First Search (BFS) and Depth-First Search (DFS).

### Python Example
```python
from collections import deque

def bfs(graph, start):
    visited = {start}
    queue = deque([start])
    while queue:
        node = queue.popleft()
        print(node, end=" ")
        for neighbor in graph[node]:
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor)

graph = {1: [2, 3], 2: [1, 4], 3: [1], 4: [2]}
bfs(graph, 1)
```

### Java Example
```java
import java.util.*;

public class Main {
    public static void main(String[] args) {
        Map<Integer, List<Integer>> graph = new HashMap<>();
        graph.put(1, Arrays.asList(2, 3));
        graph.put(2, Arrays.asList(1, 4));
        graph.put(3, Arrays.asList(1));
        graph.put(4, Arrays.asList(2));

        Queue<Integer> queue = new LinkedList<>();
        Set<Integer> visited = new HashSet<>();
        queue.add(1);
        visited.add(1);

        while (!queue.isEmpty()) {
            int node = queue.remove();
            System.out.print(node + " ");
            for (int neighbor : graph.getOrDefault(node, Collections.emptyList())) {
                if (!visited.contains(neighbor)) {
                    visited.add(neighbor);
                    queue.add(neighbor);
                }
            }
        }
    }
}
```

## 10. Dynamic Programming

Dynamic programming solves complex problems by breaking them into overlapping subproblems and reusing results.

### Python Example
```python
def fib(n):
    if n <= 1:
        return n
    dp = [0] * (n + 1)
    dp[1] = 1
    for i in range(2, n + 1):
        dp[i] = dp[i - 1] + dp[i - 2]
    return dp[n]

print(fib(8))
```

### Java Example
```java
public class Main {
    public static int fib(int n) {
        if (n <= 1) return n;
        int[] dp = new int[n + 1];
        dp[1] = 1;
        for (int i = 2; i <= n; i++) {
            dp[i] = dp[i - 1] + dp[i - 2];
        }
        return dp[n];
    }

    public static void main(String[] args) {
        System.out.println(fib(8));
    }
}
```

## 11. Greedy Algorithms

Greedy algorithms make locally optimal choices hoping they lead to a global optimum.

### Python Example
```python
def make_change(coins, amount):
    count = 0
    for coin in sorted(coins, reverse=True):
        count += amount // coin
        amount %= coin
    return count

print(make_change([1, 5, 10, 25], 41))
```

### Java Example
```java
import java.util.*;

public class Main {
    public static int makeChange(int[] coins, int amount) {
        int count = 0;
        Arrays.sort(coins);
        for (int i = coins.length - 1; i >= 0; i--) {
            count += amount / coins[i];
            amount %= coins[i];
        }
        return count;
    }

    public static void main(String[] args) {
        int[] coins = {1, 5, 10, 25};
        System.out.println(makeChange(coins, 41));
    }
}
```

## 12. Backtracking

Backtracking explores possibilities recursively and abandons paths that cannot lead to a solution.

### Python Example
```python
def backtrack(path, remaining):
    if not remaining:
        print(path)
        return
    for choice in remaining:
        next_remaining = [x for x in remaining if x != choice]
        backtrack(path + [choice], next_remaining)

backtrack([], [1, 2, 3])
```

### Java Example
```java
public class Main {
    public static void backtrack(int[] path, int depth, int[] remaining) {
        if (depth == remaining.length) {
            for (int i = 0; i < depth; i++) System.out.print(path[i] + " ");
            System.out.println();
            return;
        }
        for (int i = 0; i < remaining.length; i++) {
            int[] nextPath = path.clone();
            nextPath[depth] = remaining[i];
            backtrack(nextPath, depth + 1, remaining);
        }
    }

    public static void main(String[] args) {
        backtrack(new int[3], 0, new int[]{1, 2, 3});
    }
}
```

## 13. Heaps and Priority Queues

Heaps provide efficient access to the minimum or maximum element and are often used in priority queues.

### Python Example
