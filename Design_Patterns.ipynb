# Design Patterns in Python

This notebook introduces the classic design patterns, explains their purpose, and shows where they are useful in modern software development.

## Why Design Patterns Matter

Design patterns are reusable solutions to recurring software design problems. They help teams build systems that are:
- easier to maintain
- easier to extend
- more readable
- more flexible when requirements change

## Design Principles Behind the Patterns

Most patterns are based on a few core ideas:
- Encapsulate what changes
- Favor composition over inheritance
- Program to interfaces or abstractions
- Open for extension, closed for modification
- Depend on abstractions, not concrete implementations

## 1. Creational Patterns

Creational patterns deal with object creation. They help make systems more flexible by controlling how objects are instantiated.

### Factory Pattern
- Used when you want to create objects without exposing the exact class
- Common in plug-in systems, service creation, and object selection based on input

### Abstract Factory Pattern
- Used when a family of related objects must be created together
- Common in UI toolkits, cross-platform apps, and database abstraction layers

### Builder Pattern
- Used when an object has many optional parameters or complex construction steps
- Common in configuration objects, API clients, and report generators

### Prototype Pattern
- Used when copying an existing object is better than rebuilding it from scratch
- Common in cloning objects and document-like structures

### Singleton Pattern
- Used when exactly one shared instance should exist
- Common in logging, configuration, and shared connection pools

## 2. Structural Patterns

Structural patterns deal with how objects and classes are composed into larger structures.

### Adapter Pattern
- Used when two incompatible interfaces must work together
- Common in legacy integrations and third-party SDK wrappers

### Bridge Pattern
- Used to separate abstraction from implementation
- Common in UI frameworks and cross-platform rendering systems

### Composite Pattern
- Used when objects need to be treated uniformly as part of a tree structure
- Common in file systems, menus, and organizational hierarchies

### Decorator Pattern
- Used to add behavior dynamically without changing the original class
- Common in logging, caching, validation, and authentication

### Facade Pattern
- Used to simplify a complex subsystem behind a clean interface
- Common in service layers, SDK abstractions, and APIs

### Flyweight Pattern
- Used to reduce memory usage by sharing common state
- Common in text rendering, game objects, and large data representations

### Proxy Pattern
- Used to control access to another object
- Common in lazy loading, access control, and remote services

## 3. Behavioral Patterns

Behavioral patterns focus on communication between objects and how responsibilities are distributed.

### Strategy Pattern
- Used when multiple behaviors or algorithms can be selected at runtime
- Common in payment systems, routing logic, and sorting strategies

### Observer Pattern
- Used when one object needs to notify many dependents
- Common in event systems, UI frameworks, and pub/sub architectures

### Command Pattern
- Used to encapsulate requests as objects
- Common in undo/redo systems, task queues, and workflow engines

### State Pattern
- Used when an object behaves differently based on its state
- Common in order processing, ticketing systems, and UI states

### Template Method Pattern
- Used when an algorithm is shared but some steps vary
- Common in framework hooks, data pipelines, and libraries

### Chain of Responsibility
- Used when multiple handlers may process a request in sequence
- Common in middleware, validation chains, and logging pipelines

### Iterator Pattern
- Used to traverse collections without exposing their internal structure
- Common in custom containers and stream processing

### Mediator Pattern
- Used to reduce coupling between many objects
- Common in chat systems and complex UI interactions

### Memento Pattern
- Used to capture and restore an object's previous state
- Common in editors, undo systems, and workflow rollback

### Visitor Pattern
- Used to add operations to a set of objects without changing their classes
- Common in compilers, syntax trees, and document processing

### Interpreter Pattern
- Used for parsing and evaluating a small language
- Common in query languages, expression evaluators, and DSLs

## 4. When to Use Which Pattern

- Use Factory/Builder when object creation is complex or varies by context
- Use Adapter/Facade when integrating with external systems
- Use Decorator when you need to add behavior dynamically
- Use Strategy when behavior must change at runtime
- Use Observer when many components need to react to changes
- Use State when an object has multiple lifecycle states
- Use Command when you want to queue, log, or undo actions
